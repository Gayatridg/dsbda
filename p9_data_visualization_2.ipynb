```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
import seaborn as sns
titanic = sns.load_dataset("titanic")
```

```python
titanic
```

```python
titanic.isnull().sum()
```

```python
titanic['age'] = titanic['age'].fillna(np.mean(titanic['age'])) 

titanic['deck'] = titanic['deck'].fillna(titanic['deck'].mode()[0])
     
titanic['embark_town'] = titanic['embark_town'].fillna(titanic['embark_town'].mode()[0])
```

```python
plt.figure(figsize=(15,10))
sns.boxplot(data=titanic, x="sex", y="age", hue="alive")
```

```python
sns.boxplot(x="sex",y="age",data=titanic)
```

```python
#optional tula krycha asel tr kr
plt.figure(figsize=(8,4)) # 8 by 4 inches
sns.boxplot(x="sex", y="age", hue="survived", data= titanic, palette="viridis")
plt.title("Distribution of age with respect to each gender and survival Status")
plt.xlabel("Sex")
plt.ylabel("Age")
plt.show()
```

```python
#optionla
sns.catplot(x="sex", hue="survived", data=titanic, kind="count")
```

```python
#optional
sns.violinplot(x='sex',y='age',data=titanic, hue= 'survived')
```
