```python
df['Classification'] = df['Classification'].astype('category')
```

```python
# Replace 'na' with actual NaN
df['Classification'] = df['Classification'].replace('na', np.nan)

# Drop rows with missing classification
df = df.dropna(subset=['Classification'])
```

```python

df.loc[:, 'Classification'] = df['Classification'].astype(int)
```

```python
# Step 5: (Optional) Convert categorical columns to numeric
df_encoded = pd.get_dummies(df, drop_first=True)

# Step 6: View the cleaned data
print(df_encoded.head())
```
