```python
#optional.....................
setosa = data['species'] == 'setosa'
virginica = data['species'] == 'virginica'
versicolor = data['species'] == 'versicolor'
# Displaying the statistical details for each species
print("Statistics for Iris-setosa:")
print(setosa)

print("\nStatistics for Iris-versicolor:")
print(versicolor)

print("\nStatistics for Iris-virginica:")
print(virginica)
```

```python
# Calculate percentiles, mean, and standard deviation
percentiles = data.groupby('species').quantile([0.25, 0.5, 0.75])
means = data.groupby('species').mean()
std_devs = data.groupby('species').std()

print("\nPercentiles:")
print(percentiles)

print("\nMean Values:")
print(means)

print("\nStandard Deviations:")
print(std_devs)
```
