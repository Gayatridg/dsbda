```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
titanic = sns.load_dataset("titanic")
```

```python
titanic
```

```python
titanic.describe()
```

```python
titanic.isnull().sum()
```

```python
titanic['age'] = titanic['age'].fillna(np.mean(titanic['age'])) 

titanic['deck'] = titanic['deck'].fillna(titanic['deck'].mode()[0])
     
titanic['embark_town'] = titanic['embark_town'].fillna(titanic['embark_town'].mode()[0])

#or
#titanic_cleaned=titanic.drop(['pclass','embarked','deck','embark_town'],axis=1)
#titanic_cleaned.head(15)
```

```python
sns.histplot(data=titanic,x="fare",bins=20,binwidth=10)
```

```python
plt.figure(figsize=(6,4))
sns.displot(titanic['age']) # Use sns.distplot(df['age']) for older versions of seaborn library
plt.show()
```

```python
plt.figure(figsize=(5,3))
bp = sns.boxplot(x='class',y='age',palette='pastel',data=titanic)
plt.show()
titanic.describe().transpose
```

```python
# Histogram of Fare
plt.figure(figsize=(10, 6))
sns.histplot(data=titanic, x='fare', bins=30, kde=True)
plt.title("Fare Distribution of Titanic Passengers")
plt.xlabel("Fare")
plt.ylabel("Number of Passengers")
plt.show()
```

```python
# Example: Check class distribution
sns.countplot(x='class', hue='survived', data=titanic)
plt.title("Survival Count by Class")
plt.show()
```

```python
sns.histplot(data=titanic, x='age', hue='survived', multiple='stack')
plt.title("Age Distribution by Survival")
plt.show()
```
